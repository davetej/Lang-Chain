In Memory saver - keep full converstaion history

1. int memeory saver
2. create agent with memory saver
3. run agent with memory saver + message  + config

```python
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from langgraph.checkpoint.memory import InMemorySaver
```

```python
import os 
from dotenv import load_dotenv
load_dotenv()
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
MODEL = os.getenv("CHAT_MODEL")
```

```python
llm = init_chat_model(model=MODEL)
```

```python
checkpointer = InMemorySaver()
```

```python
chatbot = create_agent(
    model=llm,
    checkpointer=checkpointer
)
```

```python
config = {
    "configurable": {
        "thread_id": "user-1"
    }
}
```

```python
response = chatbot.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "My name is Sara."
            }
        ]
    },
    config=config
)
```

```python
from uuid import uuid4

thread_id = f"user-{uuid4().hex}"

config = {
    "configurable": {
        "thread_id": thread_id
    }
}

print(f"Conversation ID: {thread_id}")
print("Type 'exit' to stop.")

while True:
    user_input = input("\nYou: ").strip()

    if user_input.lower() in {"exit", "quit"}:
        print("Chatbot stopped.")
        break

    if not user_input:
        continue

    response = chatbot.invoke(
        {
            "messages": [
                {
                    "role": "user",
                    "content": user_input
                }
            ]
        },
        config=config
    )

    print("\nChatbot:", response["messages"][-1].content)

    state = chatbot.get_state(config)

    print("\nStored messages:")

    for message in state.values["messages"]:
        print(f"{message.type}: {message.content}")
```
