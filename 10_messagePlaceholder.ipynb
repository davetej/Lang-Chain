```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
```

```python
# ✅ Create a slot for dynamic messages
chat_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant"),
    MessagesPlaceholder(variable_name="chat_history"),  # ← Dynamic slot
    ("user", "{question}")
])
```

```python
from langchain_core.messages import HumanMessage, AIMessage
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
load_dotenv()
```

```python
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
MODEL = os.getenv("CHAT_MODEL")
```

```python
llm = init_chat_model(model=MODEL)
```

```python
chat_history = []
while True:
    # Get user input
    user_input = input("User: ")

    if user_input.lower() in ["exit", "quit"]:
        print("Exiting...")
        break

    # ✅ Generate prompt with dynamic chat history (BEFORE adding current message)
    prompt = chat_template.invoke({
        "chat_history": chat_history,
        "question": user_input
    })

    # ✅ Get model response
    response = llm.invoke(prompt)
    print(f"Assistant: {response.content}\n")

    # ✅ Add BOTH user and AI messages to history for next turn
    chat_history.append(HumanMessage(content=user_input))
    chat_history.append(AIMessage(content=response.content))
```
