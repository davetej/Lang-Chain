## Prompt Template

````python
from langchain_core.prompts import PromptTemplate

# 2. PROMPT TEMPLATE: Reusable prompts with variables
# Use when you need to plug in different values

# Create a template with a placeholder
template = PromptTemplate.from_template(
    "Explain {topic} in simple terms for a beginner {A} {B}"
)

prompt_text = template.format(topic="quantum computing", A="Alice", B="Bob")
print(type(prompt_text))
````

````python
prompt_template = template.invoke({"topic": "quantum computing", "A": "Alice", "B": "Bob"})
print(prompt_template)
print(type(prompt_template))
````

````python
#  Legacy Ignore use invoke.
prompt_template = template.format_prompt(topic="quantum computing", A="Alice", B="Bob")
print(prompt_template)
print(type(prompt_template))
````

## Chat PromptTemplate: Reusable prompts with variables for chat models

````python
from langchain_core.prompts import ChatPromptTemplate

# 3. CHAT PROMPT TEMPLATE: For conversation-style interactions
# Use when you need system instructions + user messages

chat_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {role}. Keep answers under 50 words."),
    ("user", "Explain {topic} to me")
])

prompt = chat_template.invoke({
    "role": "Python tutor",
    "topic": "list comprehensions"
})
print(prompt)
print(type(prompt))
````

````python
prompt = chat_template.format(role="Python tutor", topic="list comprehensions")
print(prompt)
print(type(prompt))
````

### Few shot Prompting

````python
from langchain_core.prompts import FewShotPromptTemplate

# 4. FEW-SHOT PROMPT TEMPLATE: Teach by showing examples
# Use when you want the AI to learn a pattern from examples

# Step 1: Define your examples
examples = [
    {"word": "happy", "emoji": "😊"},
    {"word": "sad", "emoji": "😢"},
    {"word": "angry", "emoji": "😠"},
    {"word": "excited", "emoji": "🎉"}
]
````

````python
example_template = PromptTemplate.from_template(
    "Word: {word}\nEmoji: {emoji}"
)
````

````python
few_shot = FewShotPromptTemplate(
    examples=examples,
    example_prompt=example_template,  # ← Uses PromptTemplate
    suffix="Word: {input}\nEmoji:",
    input_variables=["input"]
)
````

````python
# Step 5: Send to model and get response
few_shot_prompt = few_shot.format(input="joyful")
print(few_shot_prompt)
print(type(few_shot_prompt))
````
